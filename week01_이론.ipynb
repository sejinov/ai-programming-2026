## GitHub 사용법

### Git & GitHub

Git
- 파일의 변경 이력을 관리하는 버전 관리 도구
- 코드의 변경 시점과 내용을 기록
- 필요할 경우 이전 버전으로 복원 가능

GitHub
- Git으로 관리하는 프로젝트를 온라인에 저장·공유하는 서비스
- 코드 및 변경 이력 관리
- 다른 사용자와의 협업 지원

### GitHub를 사용하는 이유

- 버전 관리: 코드의 변경 내역 기록 및 이전 버전 복원
- 협업: 여러 사용자가 동일한 프로젝트를 함께 관리
- 과제 및 프로젝트 관리: 제출 시점과 수정 이력 확인
- 코드 공유: 프로젝트를 온라인에 저장하고 다른 사용자와 공유

### 저장소(repository)

하나의 프로젝트에 속한 파일과 변경 이력을 저장하는 공간으로, \
일반적으로 과제나 프로젝트 단위로 하나의 저장소를 구성함

로컬 저장소 (Local Repository)
- 내 컴퓨터 또는 Colab에 존재하는 저장소
- 파일 작성 및 코드 수정 작업 수행

원격 저장소 (Remote Repository)
- GitHub 서버에 존재하는 저장소
- 코드 및 변경 이력 저장·공유

로컬에서 작업한 내용을 원격으로 올리고(push), 원격의 최신 내용을 로컬로 내려받으며(pull) 두 저장소의 상태를 동기화시킴

### 기본 작업 흐름

Git의 기본적인 작업 순서

1. clone: 원격 저장소를 내 컴퓨터로 복제해 로컬 저장소 생성 (처음 한번)
2. 수정: 기존 파일 수정 또는 새 파일 작성
3. add: 변경 사항을 스테이징 영역에 추가 (커밋할 대상을 고르는 단계)
4. commit: 스테이징한 변경 내용을 하나의 버전으로 기록
5. push: 로컬의 커밋을 원격 저장소에 업로드

clone → 수정 → add → commit → push 를 반복하고, 필요할 때 pull로 최신 상태 반영

```python
# 로컬 컴퓨터의 터미널에서 실행하는 Git 명령 예시
# Colab 코드 셀에서 실행하려면 각 줄 맨 앞에 ! 를 붙이고 # 을 삭제

# 원격 저장소를 내 컴퓨터로 복제 (처음 한 번)
# git clone https://github.com/USER/REPO.git

# 파일을 수정한 뒤, 기록할 대상을 스테이징에 추가
# git add hello.py

# 스테이징한 변경을 하나의 버전으로 기록
# git commit -m "add hello.py"

# 로컬의 커밋을 원격 저장소로 업로드
# git push

# 원격의 최신 변경을 내 컴퓨터로 내려받기
# git pull
```

### Colab에서 GitHub 사용하기

Colab에서 GitHub를 사용하는 방법 두 가지:

첫째, 노트북 자체를 GitHub에 저장하기 \
Colab 메뉴에서 파일 > GitHub에 사본 저장을 선택하면, 지금 보고 있는 .ipynb 노트북을 내 GitHub 저장소에 바로 업로드 가능

둘째, 저장소의 코드를 가져오기 \
git clone으로 원격 저장소를 Colab으로 복제할 수 있으며, Colab의 코드 셀에서는 셸 명령 앞에 느낌표(!)를 붙여 실행 (ex. !git clone)

```python
# Colab에서 공개 저장소의 코드를 가져오기 (앞의 # 을 지우고 실행)
# !git clone https://github.com/USER/REPO.git

# 복제한 폴더로 이동한 뒤 파일 목록 확인
# %cd REPO
# !ls
```

### 개인 액세스 토큰(Personal Access Token)

- 공개 저장소의 clone은 별도 인증 없이 가능
- 비공개 저장소의 접근이나 원격 저장소에 push할 경우 사용자 인증 필요
- GitHub에서는 계정 비밀번호 대신 개인 액세스 토큰(Personal Access Token, PAT) 사용
- GitHub의 Settings > Developer settings > Personal access tokens에서 발급 가능

### 주의

노트북이나 코드에 토큰을 직접 작성하지 않도록 주의하고, GitHub 저장소에 토큰이 포함되지 않도록 확인할 것

```python
# 비공개 저장소를 복제하거나 push 할 때 사용하는 인증 방법으로, PAT을 발급받아 사용
# 토큰은 비밀번호처럼 취급하고, 코드에 그대로 적어 공유하지 않도록 주의

# 토큰을 URL에 포함해 복제하는 형태 (TOKEN 자리에 발급받은 토큰)
# !git clone https://TOKEN@github.com/USER/REPO.git

# Colab에서 커밋 작성자 정보를 설정한 뒤 push 하는 흐름
# !git config --global user.email "you@example.com"
# !git config --global user.name "Your Name"
# !git add .
# !git commit -m "practice week01"
# !git push https://TOKEN@github.com/USER/REPO.git
```
