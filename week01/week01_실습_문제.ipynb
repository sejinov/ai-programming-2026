## 실습 문제 5: 개인 액세스 토큰과 명령 순서

로컬에서 고친 내용을 원격 저장소에 올리려면 add, commit, push 를 순서대로 거쳐야 합니다. 비공개 저장소에 접근하거나 원격에 push 할 때는 사용자 인증이 필요하고, GitHub는 계정 비밀번호 대신 개인 액세스 토큰(Personal Access Token, PAT)을 받습니다.

- **요구사항:**
    1. Settings > Developer settings > Personal access tokens 에서 `repo` 권한을 가진 토큰을 하나 발급하세요. 토큰 값은 생성 직후에만 보이므로 노트북이 아닌 안전한 곳에 옮겨 적습니다.
